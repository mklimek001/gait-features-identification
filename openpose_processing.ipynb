```python
import os
import subprocess


sequence_key = "p4s1"
openpose_path = r"./openpose"   # change this to your OpenPose folder
video_folder = f"./gait3d/Sequences/{sequence_key}/images"      # folder containing .avi files
output_folder = r"./openpose_output"

os.makedirs(output_folder, exist_ok=True)

for file in os.listdir(video_folder):
    if file.lower().endswith(".avi"):
        input_path = os.path.join(video_folder, file)
        video_name = os.path.splitext(file)[0]

        json_out = os.path.join(output_folder, video_name + "_json")

        os.makedirs(json_out, exist_ok=True)

        cmd = [
            os.path.join(openpose_path, "bin", "OpenPoseDemo.exe"),
            "--video", input_path,
            "--write_json", json_out,
            "--display", "0"
        ]

        print("Running:", " ".join(cmd))
        subprocess.run(cmd)

print("All videos processed!")
```
