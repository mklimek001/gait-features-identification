```python
import json
import logging
from pathlib import Path
from utils.gait_parameters_extractor_raw_2d import CoordinatesIdx2D, GaitParametersExtractorRaw2D
from utils.cross_validation_siamese_raw_2d import CrossValidationSiameseRaw2D

with open("./datasets/yolo/sequences_2d/steps_camera_1.json", "r") as file:
    steps_camera_1 = json.load(file)

with open("./datasets/yolo/sequences_2d/steps_camera_3.json", "r") as file:
    steps_camera_3 = json.load(file)

with open("./datasets/yolo/sequences_2d/steps_camera_1_smoothed.json", "r") as file:
    steps_camera_1_smoothed = json.load(file)

with open("./datasets/yolo/sequences_2d/steps_camera_3_smoothed.json", "r") as file:
    steps_camera_3_smoothed = json.load(file)

with open("./datasets/yolo/selected_joint_names_v2.json", "r") as file:
    selected_joint_names = json.load(file)
```

```python
cv = CrossValidationSiameseRaw2D(
    sequence_cycles=steps_camera_1, 
    selected_joints=selected_joint_names,
    log_level=logging.DEBUG,
    logger_name = 'steps_camera_1',
)
cv.count_participants_samples()
cv.perform_training(
    n_folds = 5,
    siamese_nn_type = "conv1d",
    learning_rate = 1e-3,
    batch_size = 32,
    n_epochs = 10,
    threshold = 0.5,
    show_plot = True, 
)
```

```python
cv.perform_training_with_rank_classification(
    n_folds = 5,
    siamese_nn_type = "conv1d",
    learning_rate = 1e-3,
    batch_size = 32,
    n_epochs = 10,
    embedding_size = 10,
    tsne_perplexity = 20,
    show_plot = True, 
)
```
