```python
result = {}

for sequence_key, sequence_data in yolo_output.items():
    print(sequence_key, end=' | ')
    sequence_result = [[] for _ in range(len(sequence_data['c1'].keys()))]
    for c_idx in range(1, 5):
        camera = f"c{c_idx}"
        for f_idx, frame in sequence_data[camera].items():
            frame_scaled = width_height_multiplier * np.array(frame)
            if int(f_idx) < len(sequence_result): # p30s8 error?
                sequence_result[int(f_idx)].append(frame_scaled)

    result[sequence_key] = {i: np.array([combined_frame_result]) for i, combined_frame_result in enumerate(sequence_result)}
    
```
