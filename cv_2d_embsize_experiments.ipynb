```python
import json
import logging
from pathlib import Path
from utils.gait_parameters_extractor_raw_2d import CoordinatesIdx2D, GaitParametersExtractorRaw2D
from utils.cross_validation_siamese_raw_2d import CrossValidationSiameseRaw2D

with open("./datasets/yolo/sequences_2d/steps_yolo26_camera_1.json", "r") as file:
    steps_camera_1 = json.load(file)

with open("./datasets/yolo/sequences_2d/steps_yolo26_camera_3.json", "r") as file:
    steps_camera_3 = json.load(file)

with open("./datasets/yolo/sequences_2d/steps_yolo26_camera_1_butterworth.json", "r") as file:
    steps_camera_1_butterworth = json.load(file)

with open("./datasets/yolo/sequences_2d/steps_yolo26_camera_3_butterworth.json", "r") as file:
    steps_camera_3_butterworth = json.load(file)

with open("./datasets/yolo/selected_joint_names_v2.json", "r") as file:
    selected_joint_names = json.load(file)
```

```python
cv = CrossValidationSiameseRaw2D(
    sequence_cycles=steps_camera_3, 
    selected_joints=selected_joint_names,
    log_level=logging.INFO,
    logger_name = 'steps_camera_3',
    involve_hands_parameters=True,
)
cv.count_participants_samples()

for emb_size in [10, 20, 50, 100, 200, 500]:
    cv.perform_training(
        n_folds = 4,
        siamese_nn_type = "conv1d",
        learning_rate = 1e-5,
        embedding_size = emb_size,
        batch_size = 32,
        n_epochs = 20,
        threshold = 0.5,
        show_plot = False, 
    )
    print(f"^^^^^^^^^ {emb_size} ^^^^^^^^^")
```

```python
for emb_size in [10, 20, 50, 100, 200, 500]:
    cv.perform_training_with_rank_classification(
        n_folds = 4,
        siamese_nn_type = "conv1d",
        learning_rate = 1e-5,
        embedding_size = emb_size,
        batch_size = 32,
        n_epochs = 20,
        tsne_perplexity = 20,
        show_plot = False, 
    )
    print(f"^^^^^^^^^ {emb_size} ^^^^^^^^^")
```

```python
for emb_size in [10, 20, 50, 100, 200, 500]:
    cv.perform_training(
        n_folds = 4,
        siamese_nn_type = "lstm",
        learning_rate = 1e-4,
        embedding_size = emb_size,
        batch_size = 32,
        n_epochs = 20,
        threshold = 0.5,
        show_plot = False, 
    )
    print(f"^^^^^^^^^ {emb_size} ^^^^^^^^^")
```
