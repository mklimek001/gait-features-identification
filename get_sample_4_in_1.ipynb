```python
from utils.video_tools import get_video_files, extract_frame
```

```python
SEQ_KEY = "p1s1"

seqence_files = get_video_files(SEQ_KEY)
seqence_files
```

```python
FRAME = 67

sample_frames = [extract_frame(avi_file, FRAME, True) for avi_file in seqence_files]
```

```python
sample_frames[0]
```

```python
from PIL import Image

for i, frame in enumerate(sample_frames):
    frame_img = Image.fromarray(frame[:-40,150:-150])
    display(frame_img)
    frame_img.save(f"sample_imgs/sample_frames_video/video_frame_c{i}_{FRAME}.png")
```

```python
from PIL import Image
from IPython.display import display

width, height = 540, 960
grid_img = Image.new('RGB', (height * 2, width * 2))

positions = [
    (0, 0),
    (0, width),
    (height, 0),
    (height, width),
]

for frame, pos in zip(sample_frames, positions):
    grid_img.paste(Image.fromarray(frame), pos)

output_path = f"./sample_imgs/4_in_1_{SEQ_KEY}_{FRAME}.jpg"
grid_img.save(output_path)
print(f"Saved 2x2 grid image to {output_path}")

display(grid_img)
```
