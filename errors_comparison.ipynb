```python
datasets = {
    "mocap": "datasets/mocap/dataset_v3.json",
    
    "openpose": "datasets/openpose/dataset_openpose_triangulated_v2.json",
    
    "mediapipe" : "datasets/mediapipe/dataset_v4_triangulated.json",
    
    "hrnet": "datasets/mmpose/dataset_hrnet_triangulated.json",
    "rtmpose": "datasets/mmpose/dataset_rtmpose_triangulated.json",
    "vitpose": "datasets/mmpose/dataset_vitpose_triangulated.json",
    
    "movenet_lightning": "datasets/movenet/dataset_lightning_triangulated.json",
    "movenet_thunder": "datasets/movenet/dataset_thunder_triangulated.json",

    "yolo_v26": "datasets/yolo/dataset_v3_yolo26.json",
    # "yolo_v11": "datasets/yolo/dataset_v2.json",
    "yolo_v11": "datasets/yolo/dataset_v2_triangulated_v3.json",
}
```

```python
import json

with open(datasets['mocap'], 'r') as file:
    mocap_dataset = json.load(file) 

print(sequences := list(mocap_dataset.keys()))
print(joints := ['lfoot', 'rfoot', 'ltibia', 'rtibia', 
                 'lfemur', 'rfemur', 'lhumerus', 'rhumerus', 
                 'lradius', 'rradius', 'lwrist', 'rwrist'])
```

```python
list(datasets.keys())[1:]
```

```python
import numpy as np

SCALE_MULTIPLIER = 255

triang_ds_to_compare = list(datasets.keys())[1:]

errors = {aprox_type: {joint_key: [] for joint_key in joints} for aprox_type in triang_ds_to_compare}
mean_errors = {aprox_type: {} for aprox_type in triang_ds_to_compare}

for dataset_name in triang_ds_to_compare:
    print(f"  {dataset_name}")
    with open(datasets[dataset_name], 'r') as file:
        dataset_triang = json.load(file)
    for i, sequence_key in enumerate(sequences):
        for frame_mocap, frame_triang in zip(mocap_dataset[sequence_key], dataset_triang[sequence_key]):
            for joint_key in joints:
                errors[dataset_name][joint_key].append(np.linalg.norm(np.array(frame_mocap[joint_key]) - np.array(frame_triang[joint_key])[[1, 2, 0]]))

for aprox_type in triang_ds_to_compare:
    for joint_key in joints:
        temp_errors = errors[aprox_type][joint_key]
        mean_error = SCALE_MULTIPLIER * sum(temp_errors)/len(temp_errors)
        mean_errors[aprox_type][joint_key] = mean_error

mean_errors
```

```python
for estimator in ['joint name'] + triang_ds_to_compare:
    print("\\textbf{", estimator.replace('_', ''), "}", end="  &  ")
print("\\\\")



for joint in joints:
    print("\\textbf{", joint, "}", end=" & ")
    for estimator in triang_ds_to_compare:
        print(round(mean_errors[estimator][joint], 2), end="   &   ")
    print("\\\\")

metric = {estimator: 0 for estimator in datasets.keys()}

for estimator in triang_ds_to_compare:
    for joint in joints:
        metric[estimator] += mean_errors[estimator][joint]

print("\\textbf{summary}", end="  &  ")

for estimator in triang_ds_to_compare:
    print(round(metric[estimator]/len(joints), 2), end="   &   ")
print("\\\\")
```
