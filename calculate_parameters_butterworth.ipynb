```python
import json
from utils.gait_parameters_extractor_raw import (
    GaitParametersExtractorRaw,
    CoordinatesIdx,
)


datasets = {
    "mocap": "datasets/mocap/dataset_v3.json",
    
    "openpose": "datasets/openpose/dataset_openpose_triangulated_v2.json",
    
    "mediapipe" : "datasets/mediapipe/dataset_v4_triangulated.json",
    
    "hrnet": "datasets/mmpose/dataset_hrnet_triangulated.json",
    "rtmpose": "datasets/mmpose/dataset_rtmpose_triangulated.json",
    "vitpose": "datasets/mmpose/dataset_vitpose_triangulated.json",
    
    "movenet_lightning": "datasets/movenet/dataset_lightning_triangulated.json",
    "movenet_thunder": "datasets/movenet/dataset_thunder_triangulated.json",

    "yolo_v26": "datasets/yolo/dataset_v3_yolo26.json",
    # "yolo_v11": "datasets/yolo/dataset_v2.json",
    "yolo_v11": "datasets/yolo/dataset_v2_triangulated_v3.json",
}

def calculate_parameters_for_estimator(dataset_name: str):
    with open(datasets[dataset_name], 'r') as file:
        dataset_triang = json.load(file)
        
    combined_estimator_parameters = {}

    if dataset_name != "mocap":
        coordinates_idx=CoordinatesIdx()
    else:
        coordinates_idx=CoordinatesIdx(2, 0, 1)
    
    for sequence_key, sequence_joint_positions in dataset_triang.items():
        if sequence_joint_positions:
            print(f"| {sequence_key}", end = " ")
            gpe_raw = GaitParametersExtractorRaw(
                sequence_joint_positions, 
                coordinates_idx=coordinates_idx, 
                smooth_butterworth=True, 
                cutoff_frequency = 0.625, 
                order = 2,
                fs = 25,
            )
            parameters_value = gpe_raw.get_gait_parameters()
            parameters_names = gpe_raw.get_gait_parameters_names()
        
            combined_estimator_parameters[sequence_key] = {name: values.tolist() for name, values in zip(parameters_names, parameters_value)}

    destination_file = datasets[dataset_name].replace('dataset_', 'calculated_parameters_butterworth_')
    with open(destination_file, 'w') as f:
        json.dump(combined_estimator_parameters, f, indent=4)
    
```

```python
for dataset in datasets.keys():
    print(dataset)
    calculate_parameters_for_estimator(dataset)
```

```python
datasets_parameters = {
 'mocap': 'datasets/mocap/calculated_parameters_butterworth_v3.json',
    
 'openpose': 'datasets/openpose/calculated_parameters_butterworth_openpose_triangulated_v2.json',
    
 'mediapipe': 'datasets/mediapipe/calculated_parameters_butterworth_v4_triangulated.json',
    
 'hrnet': 'datasets/mmpose/calculated_parameters_butterworth_hrnet_triangulated.json',
 'rtmpose': 'datasets/mmpose/calculated_parameters_butterworth_rtmpose_triangulated.json',
 'vitpose': 'datasets/mmpose/calculated_parameters_butterworth_vitpose_triangulated.json',
    
 'movenet_lightning': 'datasets/movenet/calculated_parameters_butterworth_lightning_triangulated.json',
 'movenet_thunder': 'datasets/movenet/calculated_parameters_butterworth_thunder_triangulated.json',
    
 'yolo_v26': 'datasets/yolo/calculated_parameters_butterworth_v3_yolo26.json',
 'yolo_v11': 'datasets/yolo/calculated_parameters_butterworth_v2_triangulated_v3.json'
}

parameter_names = ['legs_angles', 'left_knee_angles', 'right_knee_angles', 'left_hip_angles', 'right_hip_angles', 
                   'left_humerus_angles', 'right_humerus_angles', 'left_elbow_angles', 'right_elbow_angles', 
                   'ankle_distances', 'knee_distances', 'elbow_distances', 'hand_distances', 
                   'center_of_gravity_height_change', 'lateral_pelvic_tilt', 'pelvis_rotation']
```

```python
import json

with open(datasets_parameters['mocap'], 'r') as file:
    mocap_parameters = json.load(file) 

mocap_parameters['p1s1'].keys()
```

```python
param_ds_to_compare = list(datasets_parameters.keys())[1:]
param_ds_to_compare
```

```python
import numpy as np

errors = {aprox_type: {parameter: np.array([]) for parameter in parameter_names} for aprox_type in param_ds_to_compare}
mean_errors = {aprox_type: {} for aprox_type in param_ds_to_compare}

for aprox_type in param_ds_to_compare:
    print(f"  {aprox_type}")
    with open(datasets_parameters[aprox_type], 'r') as file:
        triang_parameters = json.load(file)

    sequences = [seq_key for seq_key in triang_parameters.keys() if seq_key in mocap_parameters.keys()]
        
    for i, sequence_key in enumerate(sequences):
        for parameter in parameter_names:
            absolute_error = abs(np.array(mocap_parameters[sequence_key][parameter]) - np.array(triang_parameters[sequence_key][parameter]))
            errors[aprox_type][parameter] = np.concatenate((errors[aprox_type][parameter], absolute_error))
            

for aprox_type in param_ds_to_compare:
    for parameter in parameter_names:
        mean_error = np.mean(errors[aprox_type][parameter])
        mean_errors[aprox_type][parameter] = mean_error

mean_errors
```

```python
for estimator in ['parameter'] + param_ds_to_compare:
    print("\\textbf{", estimator.replace('_', ''), "}", end="  &  ")
print("\\\\")

for parameter in parameter_names:
    print("\\textbf{", parameter.replace('_', ' '), "}", end=" & ")
    for estimator in param_ds_to_compare:
        print(round(mean_errors[estimator][parameter], 2), end="   &   ")
    print("\\\\")
```

```python
angle_parameters = [parameter for parameter in parameter_names if 'angles' in parameter] + ['pelvis_rotation', 'lateral_pelvic_tilt']

for parameter in angle_parameters:
    print("\\textbf{", parameter.replace('_', ' '), "}", end=" & ")
    for estimator in param_ds_to_compare:
        print(round(mean_errors[estimator][parameter], 2), end="   &   ")
    print("\\\\")
```

```python
distance_parameter = [parameter for parameter in parameter_names if parameter not in angle_parameters]

for parameter in distance_parameter:
    print("\\textbf{", parameter.replace('_', ' '), "}", end=" & ")
    for estimator in param_ds_to_compare:
        print(round(mean_errors[estimator][parameter], 2), end="   &   ")
    print("\\\\")
```
