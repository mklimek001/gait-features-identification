```python
two_walks_ptcpts, one_walk_ptcpts = [], []


for ptcpt, count in participants.value_counts().to_dict().items():
    if count == 8:
        two_walks_ptcpts.append(ptcpt)
    else: 
        one_walk_ptcpts.append(ptcpt)

print(len(one_walk_ptcpts), one_walk_ptcpts)
print(len(two_walks_ptcpts), two_walks_ptcpts)
```

```python
import random
import math

def split_into_random_sublists(ptcpts_lst, n):
    splits = [[] for _ in range(n)]
    for idx, ptcpt in enumerate(ptcpts_lst):
        splits[idx%n].append(ptcpt)
    return splits
```

```python
split_into_random_sublists(one_walk_ptcpts, 5)
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.colors import LinearSegmentedColormap

gait_df = pd.read_csv('./datasets/gait_features_2.csv')


corr = gait_df[col_names_wo_asym].corr()
cmap = LinearSegmentedColormap.from_list(
    "brownish",
    ["#0f540b", "#f0e5d8", "#b09a80", "#54290b"]
)

plt.figure(figsize=(12, 12))
plt.imshow(corr, cmap=cmap)
plt.colorbar()
plt.xticks(range(len(corr.columns)), corr.columns, rotation=90)
plt.yticks(range(len(corr.columns)), corr.columns)

plt.title("Correlation Matrix")
plt.tight_layout()
plt.show()
```
