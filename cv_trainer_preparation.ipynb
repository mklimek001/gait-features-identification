```python
import random
import math

def split_into_random_sublists(ptcpts_lst, n):
    splits = [[] for _ in range(n)]
    for idx, ptcpt in enumerate(ptcpts_lst):
        splits[idx%n].append(ptcpt)
    return splits
```

```python
split_into_random_sublists(one_walk_ptcpts, 5)
```
