```python
import os

is_int = lambda line: line.strip().lstrip('-').isdigit()
raw_output_dir = "./results_different_embedding_size"
processed_output_dir = "./processed_diff_embedding_size"

for file in os.listdir(raw_output_dir):
    if file.endswith(".txt"):
        raw_file_path = raw_output_dir + '/' + file
        dest_file_path = processed_output_dir + '/' + file        
        with open(raw_file_path, "r") as infile, open(dest_file_path, "w") as outfile:
            for line in infile:
                if len((splitted_line := line.split(" - INFO - "))) > 1:
                    processed_line = splitted_line[-1].strip() + "\\\\"
                    outfile.write(processed_line + "\n")
                if is_int(line):
                    processed_line = " & \\multicolumn{3}{c}{\\textbf{" + line.strip() + " features}}\\\\"
                    outfile.write("\\midrule" + "\n")
                    outfile.write(processed_line + "\n")
                    outfile.write("\\textbf{Model} & \\textbf{Rank-1} & \\textbf{Rank-2} & \\textbf{Rank-3}\\\\" + "\n")
                    outfile.write("\\midrule" + "\n")
                
```
