```python
import json

with open("./datasets/mocap/dataset_mocap_cycles_split.json", "r") as file:
    mocap_cycles = json.load(file)

with open("./datasets/yolo/dataset_yolo_triang_cycles_split.json", "r") as file:
    yolo_cycles = json.load(file)

with open("./datasets/yolo/dataset_yolo_triang_best_cameras_cycles_split.json", "r") as file:
    yolo_best_cams_cycles = json.load(file)

with open("./datasets/yolo/dataset_yolo_triang_cycles_split_smoothed.json", "r") as file:
    yolo_cycles_smoothed = json.load(file)

with open("./datasets/yolo/dataset_yolo_triang_best_cameras_cycles_split_smoothed.json", "r") as file:
    yolo_best_cams_cycles_smoothed = json.load(file)
```

```python
import matplotlib.pyplot as plt
from matplotlib.pyplot import figure

figure(figsize=(16, 10), dpi=120)

sequence = 'p32s1c0'
y_1 = [mocap_cycles[sequence][i]['lfoot'][1]*255 for i in range(32)]
y_2 = [mocap_cycles[sequence][i]['rfoot'][1]*255 for i in range(32)]
x = [i for i in range(32)]

plt.plot(x, y_1, color='red')
plt.plot(x, y_2, color='darkred')


linestyle = ":"

y_1 = [yolo_cycles[sequence][i]['lfoot'][2]*255 for i in range(32)]
y_2 = [yolo_cycles[sequence][i]['rfoot'][2]*255 for i in range(32)]

plt.plot(x, y_1, color='blue', linestyle=linestyle)
plt.plot(x, y_2, color='darkblue', linestyle=linestyle)


y_1 = [yolo_best_cams_cycles[sequence][i]['lfoot'][2]*255 for i in range(32)]
y_2 = [yolo_best_cams_cycles[sequence][i]['rfoot'][2]*255 for i in range(32)]

plt.plot(x, y_1, color='green', linestyle=linestyle)
plt.plot(x, y_2, color='darkgreen', linestyle=linestyle)

linestyle = "--"

y_1 = [yolo_cycles_smoothed[sequence][i]['lfoot'][2]*255 for i in range(32)]
y_2 = [yolo_cycles_smoothed[sequence][i]['rfoot'][2]*255 for i in range(32)]

plt.plot(x, y_1, color='blue', linestyle=linestyle)
plt.plot(x, y_2, color='darkblue', linestyle=linestyle)

y_1 = [yolo_best_cams_cycles_smoothed[sequence][i]['lfoot'][2]*255 for i in range(32)]
y_2 = [yolo_best_cams_cycles_smoothed[sequence][i]['rfoot'][2]*255 for i in range(32)]

plt.plot(x, y_1, color='green', linestyle=linestyle)
plt.plot(x, y_2, color='darkgreen', linestyle=linestyle)


plt.xlabel('Frame')
plt.ylabel('Foot (ankle) height')
plt.title('Simple Plot')

plt.show()
```

```python
import matplotlib.pyplot as plt
from matplotlib.pyplot import figure

sequence = 'p32s3c0'

figure(figsize=(16, 8), dpi=120)
x = [i for i in range(32)]

colors = ["blue", "red", "green", "orange", "pink"]

gpe_raw = GaitParametersExtractorRaw(yolo_best_cams_cycles[sequence], coordinates_idx=CoordinatesIdx(0, 1, 2))
sequence_parameters = gpe_raw.get_gait_parameters()
gait_parameters_names = gpe_raw.get_gait_parameters_names()


for i in range(9, 14):
    plt.plot(x, sequence_parameters[i], color=colors[i-9], label=gait_parameters_names[i])

gpe_raw = GaitParametersExtractorRaw(yolo_best_cams_cycles_smoothed[sequence], coordinates_idx=CoordinatesIdx(0, 1, 2))
sequence_parameters = gpe_raw.get_gait_parameters()
gait_parameters_names = gpe_raw.get_gait_parameters_names()

for i in range(9, 14):
    plt.plot(x, sequence_parameters[i], color=colors[i-9], label=gait_parameters_names[i]+" (smoothed)", linestyle="--")
    

plt.xlabel('Frame')
plt.ylabel('Param value')
plt.title('Base gait features comparison (triangulation best cameras)')
plt.legend()

plt.show()
```
