```python
import plotly.graph_objects as go
import plotly.io as pio
pio.renderers.default = 'iframe'
SCALE_FACTOR = 255

for frame_back in range(0,13,4):
    frame_num = sequences[SEQ_KEY]['number_of_frames'] // 2
    frame_num -= frame_back
    SCALE_FACTOR = 255
    
    triangulation_result = combined_triangulation_results[SEQ_KEY][frame_num]
    
    x_t = [vec[0]/SCALE_FACTOR for vec in triangulation_result]
    y_t = [vec[1]/SCALE_FACTOR for vec in triangulation_result]
    z_t = [vec[2]/SCALE_FACTOR for vec in triangulation_result]
    
    x_t_lines = []
    y_t_lines = []
    z_t_lines = []
    
    for i, j in connections:
        x_t_lines += [x_t[i], x_t[j], None]
        y_t_lines += [y_t[i], y_t[j], None]
        z_t_lines += [z_t[i], z_t[j], None]
        
        
    fig = go.Figure(
        data=[
            go.Scatter3d(
                x=x_t, y=y_t, z=z_t,
                mode='markers+text',
                marker=dict(size=16, color='blue'),
                # text=labels_yolo,
                # hoverinfo='text',
                textposition="bottom center",
                name='Joints triangulation YOLO26'
            ),
            go.Scatter3d(
                x=x_t_lines, y=y_t_lines, z=z_t_lines,
                mode='lines',
                line=dict(color='royalblue', width=8),
                name='Bones triangulation YOLO26'
            ),
            ]
    )
    
    fig.update_layout(
        scene=dict(
            xaxis_title='X',
            yaxis_title='Y',
            zaxis_title='Z',
            xaxis=dict(range=[-5, 5], showticklabels=False),
            yaxis=dict(range=[-4, 6], showticklabels=False),
            zaxis=dict(range=[-0.5, 9.5], showticklabels=False),
            aspectmode='cube', 
        ),                 
        title='Motion capture and YOLO joins',
        title_x=0.5,
        width=1500,
        height=1000,
        legend=dict(
            orientation="h", 
            yanchor="top",
            y=-0.01, 
            xanchor="center",
            x=0.5,
            font=dict(
                size=14
            ),
        ),
        # legend={},
        scene_camera=dict(
            eye=dict(x=-0.8, y=-0.9, z=0.3),
            center=dict(x=0, y=0, z=-0.2),
            up=dict(x=0, y=0, z=1)
        )
    )
    
    # fig.show()
    
    fig.write_image(f"sample_imgs/sample_frames_3d/yolo_frame_{frame_num}.png", width=1500, height=1000, scale=1)
```

```python
with open("./datasets/yolo/dataset_yolo26.json", "r") as file:
    raw_yolo = json.load(file)

yolo_sequence = raw_yolo[SEQ_KEY]
```

```python
for camera in [f"c{i}" for i in range(1,5)]:
    frame_num = sequences[SEQ_KEY]['number_of_frames'] // 2
    # frame_num -= frame_back
    
    # Assuming 2D data is stored similarly [x, y]
    # If your data is 3D and you just want to project it, use vec[0] and vec[1]
    triangulation_result = yolo_sequence[camera][str(frame_num)]
    
    SCALE_FACTOR = 1
    x_t = [vec[0] / SCALE_FACTOR for vec in triangulation_result]
    y_t = [vec[1] / SCALE_FACTOR for vec in triangulation_result]
    
    x_t_lines = []
    y_t_lines = []
    
    for i, j in connections:
        # None acts as a break in the line to prevent connecting all joints in one stroke
        x_t_lines += [x_t[i], x_t[j], None]
        y_t_lines += [y_t[i], y_t[j], None]
        
    fig = go.Figure(
        data=[
            # Joint Markers
            go.Scatter(
                x=x_t, y=y_t,
                mode='markers',
                marker=dict(size=16, color='blue'),
                name='Joints YOLO26'
            ),
            # Bone Lines
            go.Scatter(
                x=x_t_lines, y=y_t_lines,
                mode='lines',
                line=dict(color='royalblue', width=6),
                name='Bones YOLO26'
            ),
        ]
    )
    
    fig.update_layout(
        xaxis_title='X',
        yaxis_title='Y',
        # Fixed range for consistency across frames
        xaxis=dict(range=[0, 1], showgrid=True, zeroline=False),
        yaxis=dict(
            showgrid=True, 
            zeroline=False,
            # autorange='reversed',
            range=[1, 0], 
        ),
        title=f'2D Pose Visualization - Frame {frame_num}',
        title_x=0.5,
        width=1500,
        height=1000,
        # plot_bgcolor='white', # Matches the clean 3D look
        legend=dict(
            orientation="h", 
            yanchor="top",
            y=-0.05, 
            xanchor="center",
            x=0.5,
            font=dict(size=14),
        )
    )
    
    # Save the 2D image
    fig.write_image(f"sample_imgs/sample_frames_2d/yolo_frame_{camera}_{frame_num}.png", width=1500, height=1000, scale=1)
    # fig.show()
```
