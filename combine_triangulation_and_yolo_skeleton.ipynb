```python
import plotly.graph_objects as go
import plotly.io as pio
pio.renderers.default = 'iframe'
# pio.renderers.default = 'notebook'

step = 4
frame_num = sequences[SEQ_KEY]['number_of_frames'] - 1
frame_num -= 40

offset = (-1.75, 0, 0)

position_bvh = [joint.PositionWorld for (joint, index, depth) in root.loadPose(start_frame + frame_num * step).layout()]

x = [vec[2] - offset[0] for vec in position_bvh[1:]]
y = [vec[0] - offset[1] for vec in position_bvh[1:]]
z = [vec[1] - offset[2] for vec in position_bvh[1:]]

x_lines = []
y_lines = []
z_lines = []

for i, j in parent_children_idxs:
    x_lines += [x[i], x[j], None]
    y_lines += [y[i], y[j], None]
    z_lines += [z[i], z[j], None]


SCALE_FACTOR = 255

triangulation_result = combined_triangulation_results[SEQ_KEY][frame_num]

x_t = [vec[0]/SCALE_FACTOR + offset[0] for vec in triangulation_result]
y_t = [vec[1]/SCALE_FACTOR + offset[1] for vec in triangulation_result]
z_t = [vec[2]/SCALE_FACTOR + offset[2] for vec in triangulation_result]

x_t_lines = []
y_t_lines = []
z_t_lines = []

for i, j in connections:
    x_t_lines += [x_t[i], x_t[j], None]
    y_t_lines += [y_t[i], y_t[j], None]
    z_t_lines += [z_t[i], z_t[j], None]
    
    
fig = go.Figure(
    data=[
        go.Scatter3d(
            x=x, y=y, z=z,
            mode='markers+text',
            marker=dict(size=5, color='blue'),
            text=labels,
            hoverinfo='text',
            textposition="bottom center",
            name='Joints BVH'),
        go.Scatter3d(
            x=x_lines, y=y_lines, z=z_lines,
            mode='lines',
            line=dict(color='royalblue', width=3),
            name='Bones BVH'),
        go.Scatter3d(
            x=x_t, y=y_t, z=z_t,
            mode='markers+text',
            marker=dict(size=5, color='red'),
            text=labels_yolo,
            hoverinfo='text',
            textposition="bottom center",
            name='Joints triangulation YOLO26'),
        go.Scatter3d(
            x=x_t_lines, y=y_t_lines, z=z_t_lines,
            mode='lines',
            line=dict(color='indianred', width=3),
            name='Bones triangulation YOLO26'),
        ]
)

fig.update_layout(
    scene=dict(
        xaxis_title='X',
        yaxis_title='Y',
        zaxis_title='Z',
        xaxis=dict(range=[-3.5, 3.5], showticklabels=False),
        yaxis=dict(range=[2.5, 9.5], showticklabels=False),
        zaxis=dict(range=[-0.3, 6.7], showticklabels=False),
        aspectmode='cube', 
    ),                 
    title='Motion capture and YOLO joins',
    title_x=0.5,
    width=700,
    height=700,
    legend=dict(
        orientation="h", 
        yanchor="top",
        y=-0.01, 
        xanchor="center",
        x=0.5,
        font=dict(
            size=14
        ),
    ),
    scene_camera=dict(
        eye=dict(x=0.8, y=0.9, z=0.5),
        center=dict(x=0, y=0, z=-0.05),
        up=dict(x=0, y=0, z=1)
    )
)

fig.show()
```
