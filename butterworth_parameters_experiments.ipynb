```python
import json
import logging
from utils.steps_sepatation import separate_gait_sequences_to_cycles
from utils.gait_parameters_extractor import CoordinatesIdx
from utils.cross_validation_siamese_raw import CrossValidationSiameseRaw

triang_data_file = "./datasets/yolo/dataset_v3_yolo26.json"

with open(triang_data_file, 'r') as file:
    triangulation_data = json.load(file)
```

```python
results_base = separate_gait_sequences_to_cycles(triangulation_data, 
                                            window_size=32, 
                                            minima_window_size=10, 
                                            smooth_butterworth=True, 
                                            # butterworth filter parameters
                                            cutoff_frequency = 0.625, 
                                            order = 2,
                                            fs = 25,
                                           )


cv = CrossValidationSiameseRaw(
    sequence_cycles = results_base, 
    coordinates_idx = CoordinatesIdx(0, 1, 2), 
    log_level = logging.DEBUG,
    logger_name = 'yolo26 results_base',
)

cv.perform_training_with_rank_classification(
    n_folds = 4,
    siamese_nn_type = "conv1d",
    learning_rate = 1e-6,
    batch_size = 32,
    n_epochs = 100,
    embedding_size = 10,
    tsne_perplexity = 20,
    show_plot = True, 
)
```

```python
import matplotlib.pyplot as plt

a = {
    'results raw': results_raw,
    'results butterworth (cf=0.625)': results_base,
    'results butterworth (cf=0.7)': results_compare_cf_07,
    'results butterworth (cf=0.8)': results_compare_cf_08,
    'results butterworth (cf=1)': results_compare_cf1,
    'results butterworth (cf=2)': results_compare_cf2,
    'results butterworth (cf=3)': results_compare_cf3, 
    'results butterworth (cf=4)': results_compare_cf4, 
}

plt.figure(figsize=(12, 9))

for i, (name, data) in enumerate(a.items()):
    
    ptcp_key = 'p1s1c1'
    i = 2
    t = range(32)
    
    joint = 'lfoot'
    ptcp_data = data[ptcp_key]
    lfoot = [frame[joint][i] for frame in ptcp_data]
    
    joint = 'rfoot'
    ptcp_data = data[ptcp_key]
    rfoot = [frame[joint][i] for frame in ptcp_data]
    
    # Original signal
    plt.subplot(3, 1, 1)
    plt.plot(t, lfoot, label=name)
    plt.title('Left foot')
    plt.xlabel('Frame')
    plt.ylabel('Height (mm)')
    plt.grid()
    plt.legend()
    
    # Filtered signal
    plt.subplot(3, 1, 2)
    plt.plot(t, rfoot, label=name)
    plt.title('Right foot')
    plt.xlabel('Frame')
    plt.ylabel('Height (mm)')
    plt.grid()
    plt.legend()
    
plt.tight_layout()
plt.show()
```
