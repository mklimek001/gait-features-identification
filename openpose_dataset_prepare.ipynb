```python
import json

MEDIAPIPE_JOINTS_NUMBER = 25
FRAME_SIZE = (960, 540)

cameras = ['c1', 'c2', 'c3', 'c4']

dataset = {}
dataset_with_confidence = {}
missed_counter = 0

for sequence_key in sequences.keys():
    print(f"| {sequence_key}", end =" ")
    sequence_cameras = {camera: [] for camera in cameras}
    sequence_cameras_with_confidence = {camera: [] for camera in cameras}
    for camera in cameras:
        openpose_camera_path = f"./openpose_output/{sequence_key}/{camera}"
        for openpose_frame_file in sorted(os.listdir(openpose_camera_path)):
            with open(f"{openpose_camera_path}/{openpose_frame_file}", "r") as file:
                data_openpose = json.load(file)

            joints = []
            joints_with_confidence = []

            if len(data_openpose['people']) > 0:
                for i in range(MEDIAPIPE_JOINTS_NUMBER):
                    joint = data_openpose['people'][0]['pose_keypoints_2d'][i*3:i*3+3]
                    joint_scaled = [joint[0]/FRAME_SIZE[0], joint[1]/FRAME_SIZE[1]]
                    joints.append(joint_scaled)
                    joints_with_confidence.append(joint_scaled + [joint[2]])
            else: 
                missed_counter += 1

            sequence_cameras[camera].append(joints)
            sequence_cameras_with_confidence[camera].append(joints_with_confidence)

    dataset[sequence_key] = sequence_cameras
    dataset_with_confidence[sequence_key] = sequence_cameras_with_confidence

with open("./datasets/openpose/dataset_openpose.json", "w") as file:
    json.dump(dataset, file, indent=4)

with open("./datasets/openpose/dataset_with_confidence_openpose.json", "w") as file:
    json.dump(dataset_with_confidence, file, indent=4)

print(f"Missed frames: {missed_counter}")
```
